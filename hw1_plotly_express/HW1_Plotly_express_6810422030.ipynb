```python
df = df_raw.iloc[2:-7].reset_index(drop=True)
header_part1 = df.iloc[0].ffill()
header_part2 = df.iloc[1].fillna('')
df.columns = [f"{a}_{b}".strip('_') for a, b in zip(header_part1, header_part2)]
df = df.iloc[2:].reset_index(drop=True)
cols_to_drop = [c for c in df.columns if pd.isna(c) or 'CHANG' in str(c) or 'YTD' in str(c)]
df = df.drop(columns=cols_to_drop)
df = df.iloc[:,:-1]
df = df[df['Country of_Nationality'] != 'Country of_Nationality']
df = df.rename(columns={'Country of_Nationality': 'Country of Nationality'})
display(df)
```

```python
df.info()
```

```python
# 1. กำหนดลิสต์ของภูมิภาคหลักและย่อย (อ้างอิงตามรูปภาพที่ 2)
major_regions = ['Asia and the Pacific', 'Europe', 'America', 'Middle East', 'Africa']

sub_regions = [
    'South-East Asia', 'North-East Asia', 'South Asia', 'Other in Asia', 'Oceania', 'Other in Oceania',
    'Northern Europe', 'Western Europe', 'Central/Eastern Europe', 'Southern/Medit. Europe', 'Other in Europe',
    'Other in America', 'Other in Middle East', 'Other in Africa'
]

# 2. สร้างคอลัมน์ Region และ Sub-Region
# ถ้าเจอคำในลิสต์ให้ดึงมาวาง ถ้าไม่ใช่ให้เป็น NaN เพื่อเตรียม ffill
df['Region'] = df['Country of Nationality'].where(df['Country of Nationality'].isin(major_regions))
df['Sub_Region'] = df['Country of Nationality'].where(df['Country of Nationality'].isin(sub_regions))

# 3. ใช้ ffill() เพื่อเติมชื่อกลุ่มลงมาให้ถึงแถวประเทศ
df['Sub_Region'] = df['Sub_Region'].fillna(df['Region'])
df['Region'] = df['Region'].ffill()
df['Sub_Region'] = df['Sub_Region'].ffill()

# 4. ลบแถวที่เป็นหัวข้อ (ที่ไม่มีข้อมูลตัวเลขจริงๆ) ออกไป
# โดยเราจะเก็บเฉพาะแถวที่ไม่ใช่ทั้ง Region และ Sub_Region
sub_regions.remove("Other in Asia")
sub_regions.remove("Other in Europe")
df_cleaned = df[~df['Country of Nationality'].isin(major_regions + sub_regions)].copy()

# 5. ตอนนี้ข้อมูลพร้อมทำ Unpivot (Melt) แล้วครับ
df_long = df_cleaned.melt(
    id_vars=['Region', 'Sub_Region', 'Country of Nationality'],
    var_name='Period',
    value_name='Value'
)

# 6. แยกปีและเดือนตามที่คุณต้องการ
df_long['Year'] = df_long['Period'].str.extract(r'(\d{4})')
df_long['Month'] = df_long['Period'].str.split('_').str[-1]

# แปลงค่าตัวเลข (ลบ comma)
df_long['Value'] = pd.to_numeric(df_long['Value'].astype(str).str.replace(',', ''), errors='coerce')

# จัดเรียงคอลัมน์ให้สวยงาม
df_final = df_long[['Region', 'Sub_Region', 'Country of Nationality', 'Year', 'Month', 'Value']]
df_final = df_final.rename(columns={'Value': 'Number of Arrivals'})
display(df_final.head(20))
df_final.to_excel('tourist_arrivals.xlsx', index=False)
```

```python
object_cols = df_final.select_dtypes(include=['object']).columns
df_final[object_cols] = df_final[object_cols].convert_dtypes()
display(df_final.info())
```

## Question 1 — What are the top 10 countries of nationality contributing the highest total number of tourist arrivals?

**Objective:** Understand the primary source markets for tourism overall.

#### Graph 1 — Top 10 Countries by Total Tourist Arrivals to Thailand (2024-2025)

```python
# 1. Group by country and sum arrivals
df_q1 = (
    df_final
    .groupby('Country of Nationality', as_index=False)['Number of Arrivals']
    .sum()
    .sort_values('Number of Arrivals', ascending=False)
    .head(10)
    .sort_values('Number of Arrivals', ascending=True)   # ascending for horizontal bar readability
)

# 2. Horizontal bar chart
fig1 = px.bar(
    df_q1,
    x='Number of Arrivals',
    y='Country of Nationality',
    orientation='h',
    title='Top 10 Countries by Total Tourist Arrivals to Thailand (2024-2025)',
    text='Number of Arrivals',
    color='Number of Arrivals',
    color_continuous_scale='Tealgrn',
)

fig1.update_traces(
    texttemplate='%{text:,.0f}',
    textposition='outside',
)

fig1.update_layout(
    xaxis_title='Total Number of Arrivals',
    yaxis_title='',
    coloraxis_showscale=False,
    height=500,
    margin=dict(l=10, r=120, t=60, b=40),
    font=dict(size=13),
)

fig1.show()
```

### 💡 Insights from Question 1

1. **China dominates overwhelmingly** — with over **11.2 million** total arrivals across 2024–2025, China is by far the largest source market for Thailand, almost **1.2× the second-place Malaysia** (~9.5M).
2. **Top 3 are all Asian neighbors** — China, Malaysia, and India together account for a massive share of total arrivals, reflecting the importance of geographic proximity and cultural/economic ties.
3. **Diverse long-haul presence** — Russia (~3.6M), the United States (~2.1M), and the United Kingdom (~2.0M) show that Thailand attracts significant long-haul tourism beyond its immediate region.
4. **South Korea and Japan** maintain strong positions (~3.4M and ~2.1M respectively), highlighting the continued strength of the North-East Asian market.
5. **Laos at #10** (~2.0M) reflects substantial cross-border traffic, likely driven by geographic proximity and shared borders with Thailand.

> **Takeaway:** Thailand's tourism strategy should prioritize maintaining the dominant Asian markets (China, Malaysia, India) while continuing to develop high-value long-haul segments (Russia, US, UK).

## Question 2 — How does the volume of tourist arrivals trend over time, and how do these temporal trends compare across different major Regions?

**Objective:** Analyze time-series trends, identifying peak seasons and comparing the temporal distributions across geographic macro-regions.

#### Graph 2A — Tourist Arrivals Trend by Month (2024 vs 2025)

```python
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df1 = df_final.groupby(['Year', 'Month'])['Number of Arrivals'].sum().reset_index()
df1['Month'] = pd.Categorical(df1['Month'], categories=month_order, ordered=True)
df1 = df1.sort_values(['Year', 'Month'])

fig = px.line(df1, x="Month", y="Number of Arrivals", color='Year',
              title='Tourist Arrivals Trend by Month (2024 vs 2025)'
              )
fig.show()
```

#### Graph 2B — Tourist Arrivals Trend Over Time by Region (2024-2025)

```python
# 1. Build a proper chronological Date column
month_map = {
    'Jan': '01', 'Feb': '02', 'Mar': '03', 'Apr': '04',
    'May': '05', 'Jun': '06', 'Jul': '07', 'Aug': '08',
    'Sep': '09', 'Oct': '10', 'Nov': '11', 'Dec': '12'
}

df_q2 = df_final.copy()
df_q2['Month_Num'] = df_q2['Month'].map(month_map)
df_q2['Date'] = pd.to_datetime(df_q2['Year'].astype(str) + '-' + df_q2['Month_Num'] + '-01')

# 2. Group by Date and Region
df_q2_agg = (
    df_q2
    .groupby(['Date', 'Region'], as_index=False)['Number of Arrivals']
    .sum()
    .sort_values('Date')
)

# 3. Multi-line time series
fig2 = px.line(
    df_q2_agg,
    x='Date',
    y='Number of Arrivals',
    color='Region',
    markers=True,
    title='Tourist Arrivals Trend Over Time by Region (2024-2025)',
    color_discrete_sequence=px.colors.qualitative.Bold,
)

fig2.update_layout(
    xaxis_title='Date',
    yaxis_title='Number of Arrivals',
    legend_title='Region',
    hovermode='x unified',
    height=550,
    font=dict(size=13),
)

fig2.update_xaxes(dtick='M1', tickformat='%b %Y')

fig2.show()
```

### 💡 Insights from Question 2

1. **Asia and the Pacific is the dominant region** — its line sits far above all other regions throughout the entire period, consistently contributing **1.5M–2.4M arrivals per month**.
2. **Clear seasonality patterns** — across all regions, there are visible seasonal peaks. For Asia, **January** and **December** tend to be high seasons (winter holidays), while **March–April** dips slightly.
3. **Europe shows a strong winter-peak pattern** — European arrivals peak around **January–March** and **December**, coinciding with European winter when travelers seek warm destinations like Thailand.
4. **2025 vs 2024 comparison** — Asia and the Pacific shows a noticeable decline in 2025 compared to 2024 in several months, suggesting possible shifts in travel behavior or policy changes impacting Chinese/Malaysian visitors.
5. **America, Middle East, and Africa remain relatively flat** — these regions contribute smaller, more stable volumes without dramatic seasonal fluctuations, indicating consistent but niche demand.

> **Takeaway:** Peak-season marketing should be region-specific: target Europeans in Q4/Q1, leverage year-round Asian demand, and investigate the 2025 decline in Asian arrivals to identify corrective actions.

## Question 3 — What is the hierarchical market share composition of tourist arrivals, flowing from Regions down to Sub-Regions and finally to specific Countries?

**Objective:** Perform a multidimensional breakdown to visualize how individual countries roll up into sub-regions and macro-regions, showing their proportional contribution to the whole.

### Market Dependency and Concentration Risk

**The Insight Issue:** Are we overly reliant on a concentrated cluster of source markets?  
High dependency introduces significant revenue risk if economic disruptions occur in those areas.

#### Graph 3A — Macro View: Regional & Sub-Regional & Country Market Share (Treemap)

```python
# 1. Aggregate by Region > Sub_Region > Country
df_q3 = (
    df_final
    .groupby(['Region', 'Sub_Region', 'Country of Nationality'], as_index=False)['Number of Arrivals']
    .sum()
)

# 2. Filter out rows where arrivals are 0
df_q3 = df_q3[df_q3['Number of Arrivals'] > 0]

# 3. Treemap
fig3 = px.treemap(
    df_q3,
    path=['Region', 'Sub_Region', 'Country of Nationality'],
    values='Number of Arrivals',
    color='Number of Arrivals',
    color_continuous_scale='RdYlGn',
    title='Hierarchical Market Share: Region → Sub-Region → Country (2024-2025)',
)

fig3.update_layout(
    height=700,
    font=dict(size=13),
    coloraxis_colorbar=dict(title='Arrivals'),
)

fig3.update_traces(
    textinfo='label+value+percent parent',
    hovertemplate='<b>%{label}</b><br>Arrivals: %{value:,.0f}<br>Share of parent: %{percentParent:.1%}<extra></extra>',
)

fig3.show()
```

### 💡 Insights from Graph 3A

1. **Asia and the Pacific is the overwhelming majority** — this macro-region accounts for the largest block in the treemap, confirming that Thailand's tourism is heavily Asia-centric.
2. **Within Asia, South-East Asia and North-East Asia are the two largest sub-regions** — South-East Asia (driven by Malaysia, Laos, Vietnam) and North-East Asia (driven by China, South Korea, Japan, Taiwan) form the core pillars of arrivals.
3. **China alone** forms one of the single largest country-level blocks, visually dominating even within the already-large North-East Asia sub-region — a sign of extreme market concentration.
4. **Europe is the clear second tier** — within Europe, Northern Europe (led by the UK, Sweden, Finland), Western Europe (Germany, France, Netherlands), and Southern/Mediterranean Europe (Italy, Spain) are well-balanced sub-regions.
5. **Africa and Middle East are small slivers** — showing minimal proportional contribution, but individual countries like South Africa and Israel/UAE are identifiable within them.
6. **Hierarchical drill-down reveals hidden structure** — while the bar chart (Q1) shows top individual countries, the treemap reveals *why* certain regions dominate: it is not just one large country, but a cluster of contributing nations within each sub-region.

> **Takeaway:** The treemap highlights a **concentration risk** — Thailand's tourism depends heavily on a few Asian sub-regions. Diversifying into under-represented markets (Africa, Middle East, Americas) could build a more resilient tourism portfolio.

#### Graph 3B — Micro / Dependency View: Top 15 Countries Pareto Chart

```python
# Aggregate top 15 countries
df_3b = (
    df_final
    .groupby('Country of Nationality', as_index=False)['Number of Arrivals']
    .sum()
    .sort_values('Number of Arrivals', ascending=False)
    .head(15)
    .reset_index(drop=True)
)
total_arrivals = df_final['Number of Arrivals'].sum()
df_3b['Cumulative %'] = df_3b['Number of Arrivals'].cumsum() / total_arrivals * 100

# Build Pareto chart: bars + cumulative line
fig1b = go.Figure()

fig1b.add_trace(go.Bar(
    x=df_3b['Country of Nationality'],
    y=df_3b['Number of Arrivals'],
    name='Arrivals',
    marker_color='steelblue',
    text=df_3b['Number of Arrivals'].apply(lambda v: f'{v/1e6:.1f}M'),
    textposition='outside',
    yaxis='y1',
))

fig1b.add_trace(go.Scatter(
    x=df_3b['Country of Nationality'],
    y=df_3b['Cumulative %'],
    name='Cumulative %',
    mode='lines+markers',
    line=dict(color='firebrick', width=2.5),
    marker=dict(size=7),
    yaxis='y2',
))

# Reference lines at 50% and 80%
for pct, color, dash in [(50, 'orange', 'dash'), (80, 'green', 'dot')]:
    fig1b.add_hline(
        y=pct, line_dash=dash, line_color=color, line_width=1.5,
        annotation_text=f'{pct}%', annotation_position='right',
        yref='y2',
    )

fig1b.update_layout(
    title='Pareto Chart: Top 15 Countries by Total Arrivals (2024–2025)',
    xaxis=dict(title='Country', tickangle=-30),
    yaxis=dict(title='Number of Arrivals', showgrid=True),
    yaxis2=dict(title='Cumulative %', overlaying='y', side='right',
                range=[0, 105], ticksuffix='%', showgrid=False),
    legend=dict(x=0.9, y=-0.5),
    height=520,
    font=dict(size=13),
    hovermode='x unified',
)
fig1b.show()
```

### 💡 Key Takeaways: Market Dependency & Concentration Risk

**Concentration Ratio (Pareto Analysis):**
- **Just 6 countries** (China, Malaysia, India, Russia, South Korea, Japan) account for the first **50%** of all arrivals.
- The top **15 countries** collectively account for approximately **75%** of total arrivals — a classic Pareto concentration pattern.

**Vulnerability:**
- **China alone contributes ~16.4%** of total arrivals — the single largest source market. Any economic shock, policy change, or travel restriction in China would have an outsized impact on Thailand's tourism revenue.
- The top 2 countries (China + Malaysia) together represent over **30%** of all global arrivals.

**Diversification Opportunities (from Graph 3A Treemap):**
- **South Asia** (Bangladesh, India, Pakistan, Nepal, Sri Lanka) shows an evenly distributed mix with no single country overwhelmingly dominant — a healthier, more resilient sub-region.
- **Western Europe** (Germany, France, Netherlands, Switzerland) and **Northern Europe** (UK, Sweden, Norway) each contribute well-balanced flows that collectively rival any single Asian country.
- **Southern/Mediterranean Europe** has the highest growth momentum (+22.4% YoY) and remains an underdeveloped market relative to its potential.

> **Strategic Improvement:** Design targeted marketing campaigns in South Asia, Western Europe, and the Americas to reduce over-reliance on China and Malaysia. Even a 5% shift from Chinese arrivals into three other markets would create a structurally safer portfolio.
