```python
import pandas as pd
import plotly.express as px
import numpy as np
```

```python
file_path = '/content/DADS5001/hw1_plotly_express/International Tourist Arrivals to Thailand Jan - Dec 2025.csv'
df_raw = pd.read_csv(file_path)
df_raw
```

```python
df = df_raw.iloc[2:-7].reset_index(drop=True)
header_part1 = df.iloc[0].ffill()
header_part2 = df.iloc[1].fillna('')
df.columns = [f"{a}_{b}".strip('_') for a, b in zip(header_part1, header_part2)]
df = df.iloc[2:].reset_index(drop=True)
cols_to_drop = [c for c in df.columns if pd.isna(c) or 'CHANG' in str(c) or 'YTD' in str(c)]
df = df.drop(columns=cols_to_drop)
df = df.iloc[:,:-1]
df = df[df['Country of_Nationality'] != 'Country of_Nationality']
df = df.rename(columns={'Country of_Nationality': 'Country of Nationality'})
display(df)
```

```python
df.info()
```

```python
# 1. กำหนดลิสต์ของภูมิภาคหลักและย่อย (อ้างอิงตามรูปภาพที่ 2)
major_regions = ['Asia and the Pacific', 'Europe', 'America', 'Middle East', 'Africa']

sub_regions = [
    'South-East Asia', 'North-East Asia', 'South Asia', 'Other in Asia', 'Oceania', 'Other in Oceania',
    'Northern Europe', 'Western Europe', 'Central/Eastern Europe', 'Southern/Medit. Europe', 'Other in Europe',
    'Other in America', 'Other in Middle East', 'Other in Africa'
]

# 2. สร้างคอลัมน์ Region และ Sub-Region
# ถ้าเจอคำในลิสต์ให้ดึงมาวาง ถ้าไม่ใช่ให้เป็น NaN เพื่อเตรียม ffill
df['Region'] = df['Country of Nationality'].where(df['Country of Nationality'].isin(major_regions))
df['Sub_Region'] = df['Country of Nationality'].where(df['Country of Nationality'].isin(sub_regions))

# 3. ใช้ ffill() เพื่อเติมชื่อกลุ่มลงมาให้ถึงแถวประเทศ
df['Sub_Region'] = df['Sub_Region'].fillna(df['Region'])
df['Region'] = df['Region'].ffill()
df['Sub_Region'] = df['Sub_Region'].ffill()

# 4. ลบแถวที่เป็นหัวข้อ (ที่ไม่มีข้อมูลตัวเลขจริงๆ) ออกไป
# โดยเราจะเก็บเฉพาะแถวที่ไม่ใช่ทั้ง Region และ Sub_Region
sub_regions.remove("Other in Asia")
sub_regions.remove("Other in Europe")
df_cleaned = df[~df['Country of Nationality'].isin(major_regions + sub_regions)].copy()

# 5. ตอนนี้ข้อมูลพร้อมทำ Unpivot (Melt) แล้วครับ
df_long = df_cleaned.melt(
    id_vars=['Region', 'Sub_Region', 'Country of Nationality'],
    var_name='Period',
    value_name='Value'
)

# 6. แยกปีและเดือนตามที่คุณต้องการ
df_long['Year'] = df_long['Period'].str.extract(r'(\d{4})')
df_long['Month'] = df_long['Period'].str.split('_').str[-1]

# แปลงค่าตัวเลข (ลบ comma)
df_long['Value'] = pd.to_numeric(df_long['Value'].astype(str).str.replace(',', ''), errors='coerce')

# จัดเรียงคอลัมน์ให้สวยงาม
df_final = df_long[['Region', 'Sub_Region', 'Country of Nationality', 'Year', 'Month', 'Value']]
df_final = df_final.rename(columns={'Value': 'Number of Arrivals'})
display(df_final.head(20))
df_final.to_excel('tourist_arrivals.xlsx', index=False)
```

```python
object_cols = df_final.select_dtypes(include=['object']).columns
df_final[object_cols] = df_final[object_cols].convert_dtypes()
display(df_final.info())
```

```python
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df1 = df_final.groupby(['Year', 'Month'])['Number of Arrivals'].sum().reset_index()
df1['Month'] = pd.Categorical(df1['Month'], categories=month_order, ordered=True)
df1 = df1.sort_values(['Year', 'Month'])

fig = px.line(df1, x="Month", y="Number of Arrivals", color='Year',
              title='Monthly Trends'
              )
fig.show()
```
